```python

import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# 1. SYSTEM PARAMETERS
# ==========================================

N = 100000
snr_db = 10

np.random.seed(42)

# ==========================================
# 2. GENERATE BPSK DATA
# ==========================================

bits = np.random.randint(0, 2, N)

# BPSK modulation: 0 -> -1, 1 -> +1
x = 2 * bits - 1

# ==========================================
# 3. GENERATE RAYLEIGH FADING CHANNEL
# ==========================================

h = (np.random.randn(N) +
     1j * np.random.randn(N)) / np.sqrt(2)

# Fading envelope
envelope = np.abs(h)

# ==========================================
# 4. GENERATE AWGN NOISE
# ==========================================

snr = 10 ** (snr_db / 10)

noise_var = 1 / snr

n = np.sqrt(noise_var / 2) * (
    np.random.randn(N) +
    1j * np.random.randn(N)
)

# ==========================================
# 5. TRANSMIT THROUGH RAYLEIGH CHANNEL
# ==========================================

y = h * x + n

# ==========================================
# 6. CHANNEL EQUALIZATION
# ==========================================

y_eq = y / h

# ==========================================
# 7. BPSK DEMODULATION
# ==========================================

bits_hat = (y_eq.real > 0).astype(int)

# Calculate BER
ber = np.mean(bits_hat != bits)

print("========== SYSTEM RESULTS ==========")
print("Number of Bits =", N)
print("Modulation = BPSK")
print("SNR =", snr_db, "dB")
print("BER =", ber)

# ==========================================
# 8. TRANSMITTED AND RECEIVED SIGNAL PLOT
# ==========================================

samples = 100

plt.figure(figsize=(10, 5))

plt.plot(x[:samples], label="Transmitted BPSK")
plt.plot(y.real[:samples],
         label="Received Signal (Real Part)",
         alpha=0.8)

plt.xlabel("Sample Index")
plt.ylabel("Amplitude")
plt.title("Transmitted and Received Signals")

plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

# ==========================================
# 9. RAYLEIGH FADING ENVELOPE HISTOGRAM
# ==========================================

plt.figure(figsize=(8, 5))

plt.hist(envelope, bins=80, density=True,
         label="Rayleigh Fading Envelope")

plt.xlabel("Fading Envelope |h|")
plt.ylabel("Probability Density")
plt.title("Rayleigh Fading Envelope Histogram")

plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

# ==========================================
# 10. BER CALCULATION FOR MULTIPLE SNR
# ==========================================

snr_values = [0, 5, 10, 15, 20]

ber_values = []

for snr_db in snr_values:

    snr = 10 ** (snr_db / 10)

    noise_var = 1 / snr

    # Generate complex AWGN
    noise = np.sqrt(noise_var / 2) * (
        np.random.randn(N) +
        1j * np.random.randn(N)
    )

    # Rayleigh fading transmission
    y = h * x + noise

    # Equalization
    y_eq = y / h

    # Demodulation
    bits_hat = (y_eq.real > 0).astype(int)

    # BER
    ber = np.mean(bits_hat != bits)

    ber_values.append(ber)

    print("SNR:", snr_db, "dB",
          "BER:", ber)

# ==========================================
# 11. BER VS SNR GRAPH
# ==========================================

plt.figure(figsize=(8, 5))

# Avoid zero values on logarithmic axis
ber_plot = np.maximum(ber_values, 0.5 / N)

plt.semilogy(
    snr_values,
    ber_plot,
    marker="o",
    label="Rayleigh Fading"
)

plt.xlabel("SNR (dB)")
plt.ylabel("Bit Error Rate (BER)")
plt.title("BER vs SNR for Rayleigh Fading Channel")

plt.grid(True, which="both")
plt.legend()
plt.tight_layout()
plt.show()

# ==========================================
# 12. FINAL MESSAGE
# ==========================================

print("\nRayleigh Fading Simulation Completed Successfully!")
```

```python

import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# 1. SYSTEM PARAMETERS
# ==========================================

N = 1000000
snr_values = [0, 5, 10, 15, 20]

np.random.seed(42)

# ==========================================
# 2. GENERATE BPSK DATA
# ==========================================

bits = np.random.randint(0, 2, N)

# BPSK modulation
x = 2 * bits - 1

# ==========================================
# 3. GENERATE RAYLEIGH CHANNEL
# ==========================================

h = (np.random.randn(N) +
     1j * np.random.randn(N)) / np.sqrt(2)

# ==========================================
# 4. BER ARRAYS
# ==========================================

ber_awgn = []
ber_rayleigh = []

# ==========================================
# 5. COMPARE BOTH CHANNELS
# ==========================================

for snr_db in snr_values:

    snr = 10 ** (snr_db / 10)

    noise_var = 1 / snr

    # Generate complex AWGN
    noise = np.sqrt(noise_var / 2) * (
        np.random.randn(N) +
        1j * np.random.randn(N)
    )

    # --------------------------------------
    # AWGN ONLY CHANNEL
    # --------------------------------------

    y_awgn = x + noise

    bits_awgn = (y_awgn.real > 0).astype(int)

    ber1 = np.mean(bits_awgn != bits)

    ber_awgn.append(ber1)

    # --------------------------------------
    # RAYLEIGH FADING + AWGN CHANNEL
    # --------------------------------------

    y_rayleigh = h * x + noise

    # Channel equalization
    y_eq = y_rayleigh / h

    bits_rayleigh = (y_eq.real > 0).astype(int)

    ber2 = np.mean(bits_rayleigh != bits)

    ber_rayleigh.append(ber2)

    # --------------------------------------
    # PRINT RESULTS
    # --------------------------------------

    print("\nSNR =", snr_db, "dB")
    print("AWGN BER =", ber1)
    print("Rayleigh BER =", ber2)

# ==========================================
# 6. BER COMPARISON GRAPH
# ==========================================

plt.figure(figsize=(10, 6))

# Avoid zero values in logarithmic plot
floor = 0.5 / N

awgn_plot = np.maximum(ber_awgn, floor)
rayleigh_plot = np.maximum(ber_rayleigh, floor)

plt.semilogy(
    snr_values,
    awgn_plot,
    marker="o",
    label="AWGN Only"
)

plt.semilogy(
    snr_values,
    rayleigh_plot,
    marker="s",
    label="Rayleigh Fading + AWGN"
)

plt.xlabel("SNR (dB)")
plt.ylabel("Bit Error Rate (BER)")

plt.title("BER Comparison: AWGN vs Rayleigh Fading")

plt.grid(True, which="both")
plt.legend()
plt.tight_layout()
plt.show()

# ==========================================
# 7. FINAL RESULTS TABLE
# ==========================================

print("\n========== FINAL BER COMPARISON ==========")

print("SNR(dB)       AWGN BER       Rayleigh BER")

for i in range(len(snr_values)):

    print(
        snr_values[i],
        "          ",
        ber_awgn[i],
        "       ",
        ber_rayleigh[i]
    )

print("\nChallenge Task Completed Successfully!")
```
